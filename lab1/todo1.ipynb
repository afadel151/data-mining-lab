```python
import pandas as pd 
import numpy as np 


df = pd.read_csv("Bacteria_dataset_Multiresictance.csv",encoding='utf-8')
df.columns = df.columns.str.strip()

df["Souches"] = df["Souches"].str.strip()

df["Souches"] = df["Souches"].str.replace(r"^S\d+\s*", "", regex=True)

df["Souches"] = df["Souches"].str.replace(r"\s{2,}", " ", regex=True).str.strip()



normalization_rules = {
    # --- Escherichia coli variants ---
    r"(?i).*Escherichia\s*coli.*": "Escherichia coli",
    r"(?i).*E\.\s*coli.*": "Escherichia coli",
    r"(?i).*E\.?\s*cli.*": "Escherichia coli",
    r"(?i).*E\.?\s*coi.*": "Escherichia coli",

    # --- Klebsiella pneumoniae variants ---
    r"(?i).*Klebsie\.?lla\s*pneumoniae.*": "Klebsiella pneumoniae",
    r"(?i).*Klebsiella\s*pneumoniae.*": "Klebsiella pneumoniae",
    r"(?i).*Klbsiella\s*pneumoniae.*": "Klebsiella pneumoniae",
    
    # --- Proteus mirabilis ---
    r"(?i).*Proteus\s*mirabilis.*": "Proteus mirabilis",
    r"(?i).*Prot\.?eus\s*mirabilis.*": "Proteus mirabilis",
    r"(?i).*Proeus\s*mirabilis.*": "Proteus mirabilis",   
    r"(?i)^Protus\s*mirabilis$": "Proteus mirabilis",
    
    
    r"(?i)^Enter\.?bacteria\s*spp\.?$": "Enterobacteriaceae spp",
    r"(?i)^Enteobacteria\s*spp\.?$": "Enterobacteriaceae spp",
    r"(?i)^Enterobacteria\s*spp\.?$": "Enterobacteriaceae spp",
    r"(?i)^Enterobacteriaceae\s*spp\.?$": "Enterobacteriaceae spp"
}

for pattern, replacement in normalization_rules.items():
    df["Souches"] = df["Souches"].replace(pattern, replacement, regex=True)
    
    
df["Souches"] = df["Souches"].replace('No',pd.NA)
df["Souches"] = df["Souches"].replace('NA',pd.NA)
df["Souches"] = df["Souches"].replace('True',pd.NA)
df["Souches"] = df["Souches"].replace('None',pd.NA)
df["Souches"] = df["Souches"].replace('?',pd.NA)
df["Souches"] = df["Souches"].replace('missing',pd.NA)
df["Souches"] = df["Souches"].replace('',pd.NA)
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns if nee
print(df['Souches'].value_counts())
```
